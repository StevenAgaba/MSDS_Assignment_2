### Payroll Calculation

This section combines PAYE, NSSF and Local Service Tax to determine net pay.


```python
ctx = TaxContext(
    gross_pay=500000,
    allowances=50000,
    resident=True
)

paye = PAYE()
nssf = NSSF()
lst = LocalServiceTax()

paye_tax = paye.calculate(ctx)

nssf_result = nssf.calculate(ctx)

lst_tax = lst.calculate(ctx)

net_pay = (
    ctx.total_income
    - paye_tax
    - lst_tax
    - nssf_result["employee"]
)

print("Gross Income:", ctx.total_income)

print("PAYE:", paye_tax)

print("Local Service Tax:", lst_tax)

print("Employee NSSF:", nssf_result["employee"])

print("Employer NSSF:", nssf_result["employer"])

print("Net Pay:", net_pay)
```

## Deduction Order

The payroll calculation was performed using the following deduction sequence:

1. Local Service Tax
2. PAYE
3. Employee NSSF

The remaining balance is the employee's net pay.

## Why Float Is Unsafe

Float values may introduce rounding errors.

Example:

0.1 + 0.2 = 0.30000000000000004

Tax systems require exact arithmetic.

The Decimal class avoids such errors.


## Why Inheritance Was Used

All taxes share a common calculate() method.

An abstract Tax base class was therefore created.

PAYE, NSSF and CorporateTax inherit from Tax and provide their own implementations of calculate().
