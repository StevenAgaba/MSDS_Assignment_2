# Part A.1

Using urllib.request, urllib.parse and json to geocode a location.

```python
#Using urllib
#The request successfully converts the location name into latitude and longitude coordinates

import urllib.request
import urllib.parse
import json

location = "FortPortal"

params = urllib.parse.urlencode({
    "name": location,
    "count": 1
})

url = (
    "https://geocoding-api.open-meteo.com/v1/search?"
    + params
)

response = urllib.request.urlopen(url)

data = json.loads(response.read())

data
```

The geocoding API successfully converts a location name into latitude and longitude coordinates.

```python
#Using Requests
#Repeat the geocoding request using requests

response = requests.get(
    "https://geocoding-api.open-meteo.com/v1/search",
    params={
        "name":"FortPortal",
        "count":1
    },
    timeout=10
)

response.json()
```

Inspecting status code, content type and response time.

```python
#Inspecting status code, content type and response time.

start = time.perf_counter()

response = requests.get(
    "https://geocoding-api.open-meteo.com/v1/search",
    params={"name": "FortPortal"},
    timeout=10
)

end = time.perf_counter()

print("Status Code:", response.status_code)

print(
    "Content Type:",
    response.headers.get("Content-Type")
)

print(
    "Response Time:",
    round(end - start, 3),
    "seconds"
)
```

# Part B

Object-Oriented Weather Forecast Retrieval

```python
#Loading the class from my module
from src.weather_client import WeatherAPIClient
```

```python
client = WeatherAPIClient()
```

```python
location = "Mbale"
lat, lon = client.geocode_location(location)

print(f"Coordinates for {location}:")
print(lat, lon)
```

```python
forecast = client.get_forecast(
    lat,
    lon
)
```

Save the retrieved JSON response for offline verification.

```python
with open(
    "sample_weather.json",
    "w"
) as file:
    
    json.dump(
        forecast,
        file,
        indent=4
    )
```

Convert JSON data to a Pandas DataFrame.

```python
df = pd.DataFrame({

    "Date":
        forecast["daily"]["time"],

    "Maximum Temperature (°C)":
        forecast["daily"]["temperature_2m_max"],

    "Minimum Temperature (°C)":
        forecast["daily"]["temperature_2m_min"],

    "Rainfall (mm)":
        forecast["daily"]["precipitation_sum"]

})

df
```

Weather Data Summary Statistics

```python
print(
    "Average Maximum Temperature:",
    df["Maximum Temperature (°C)"].mean()
)

print(
    "Average Minimum Temperature:",
    df["Minimum Temperature (°C)"].mean()
)

print(
    "Total Rainfall:",
    df["Rainfall (mm)"].sum()
)
```
